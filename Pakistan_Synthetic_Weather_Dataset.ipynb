# PAKISTAN SYNTHETIC WEATHER DATASET

## 1. Introduction
The purpose of this assignment is to generate a synthetic weather dataset for 10 different cities of Pakistan using Python. The dataset contains daily weather information for one complete year, with 365 records for each city.

## 2. Selected Cities
The following 10 Pakistani cities are included:

1. Lahore
2. Karachi
3. Islamabad
4. Peshawar
5. Quetta
6. Multan
7. Faisalabad
8. Hyderabad
9. Gilgit
10. Skardu

## 3. Weather Parameters and Natural Ranges

The dataset contains seven weather parameters:

| Parameter | Unit | Approximate Range |
|---|---|---|
| Temperature | °C | Depends on city and season |
| Humidity | % | 10–100 |
| Wind Speed | km/h | 0–60 |
| Rainfall | mm | 0–60 per day |
| Atmospheric Pressure | hPa | 990–1030 |
| Visibility | km | 0.5–20 |
| UV Index | Index | 0–11 |

These are approximate synthetic modelling ranges, not actual measured weather observations. Weather values are generated using random numbers and adjusted according to city and season.

## 4. Dataset Generation Method
Python dictionaries are used to store city information and weather records. The random module generates synthetic weather values. Loops generate 365 daily weather records for each city.

## 5. Display Function
A function named `display_city_weather(city)` accepts a city name and displays all 365 weather records with appropriate column headings.

## 6. Dataset Verification
The program generates:
- 10 unique Pakistani cities.
- 365 daily records for each city.
- 3,650 total weather records.
- Seven weather parameters for each record.

## 7. Conclusion
This assignment demonstrates the use of Python dictionaries, random number generation, loops, and functions to create a synthetic weather dataset. The generated dataset can be used for basic data analysis and further machine learning practice.

```python

import random
from datetime import date, timedelta

# Reproducible random data
random.seed(42)

YEAR = 2025

# 1. City-specific climate information
# Temperature ranges are approximate synthetic assumptions.

city_climate = {

    "Lahore": {
        "temperature": {
            "winter": (5, 20), "spring": (18, 35),
            "summer": (28, 45), "autumn": (18, 35)
        },
        "rain_probability": {
            "winter": 0.15, "spring": 0.15,
            "summer": 0.35, "autumn": 0.10
        },
        "humidity_offset": 0,
        "wind_offset": 0
    },

    "Karachi": {
        "temperature": {
            "winter": (15, 30), "spring": (23, 36),
            "summer": (27, 36), "autumn": (23, 35)
        },
        "rain_probability": {
            "winter": 0.05, "spring": 0.05,
            "summer": 0.30, "autumn": 0.10
        },
        "humidity_offset": 12,
        "wind_offset": 5
    },

    "Islamabad": {
        "temperature": {
            "winter": (2, 18), "spring": (15, 32),
            "summer": (25, 40), "autumn": (15, 32)
        },
        "rain_probability": {
            "winter": 0.20, "spring": 0.20,
            "summer": 0.40, "autumn": 0.15
        },
        "humidity_offset": 8,
        "wind_offset": 0
    },

    "Peshawar": {
        "temperature": {
            "winter": (3, 20), "spring": (18, 35),
            "summer": (28, 43), "autumn": (18, 35)
        },
        "rain_probability": {
            "winter": 0.15, "spring": 0.15,
            "summer": 0.20, "autumn": 0.10
        },
        "humidity_offset": 0,
        "wind_offset": 2
    },

    "Quetta": {
        "temperature": {
            "winter": (-8, 15), "spring": (5, 28),
            "summer": (18, 38), "autumn": (5, 28)
        },
        "rain_probability": {
            "winter": 0.20, "spring": 0.15,
            "summer": 0.10, "autumn": 0.10
        },
        "humidity_offset": -10,
        "wind_offset": 3
    },

    "Multan": {
        "temperature": {
            "winter": (7, 23), "spring": (22, 38),
            "summer": (32, 47), "autumn": (22, 38)
        },
        "rain_probability": {
            "winter": 0.10, "spring": 0.10,
            "summer": 0.20, "autumn": 0.05
        },
        "humidity_offset": -5,
        "wind_offset": 0
    },

    "Faisalabad": {
        "temperature": {
            "winter": (5, 21), "spring": (19, 36),
            "summer": (29, 45), "autumn": (19, 36)
        },
        "rain_probability": {
            "winter": 0.12, "spring": 0.12,
            "summer": 0.30, "autumn": 0.08
        },
        "humidity_offset": 0,
        "wind_offset": 0
    },

    "Hyderabad": {
        "temperature": {
            "winter": (12, 29), "spring": (23, 39),
            "summer": (27, 40), "autumn": (23, 38)
        },
        "rain_probability": {
            "winter": 0.05, "spring": 0.05,
            "summer": 0.20, "autumn": 0.05
        },
        "humidity_offset": 5,
        "wind_offset": 4
    },

    "Gilgit": {
        "temperature": {
            "winter": (-10, 10), "spring": (5, 25),
            "summer": (15, 35), "autumn": (5, 25)
        },
        "rain_probability": {
            "winter": 0.10, "spring": 0.15,
            "summer": 0.20, "autumn": 0.10
        },
        "humidity_offset": -8,
        "wind_offset": 0
    },

    "Skardu": {
        "temperature": {
            "winter": (-15, 8), "spring": (0, 18),
            "summer": (10, 30), "autumn": (0, 20)
        },
        "rain_probability": {
            "winter": 0.10, "spring": 0.15,
            "summer": 0.15, "autumn": 0.10
        },
        "humidity_offset": -8,
        "wind_offset": 2
    }
}


# 2. Seasonal weather ranges

season_data = {

    "winter": {
        "humidity": (35, 75),
        "wind": (0, 25),
        "pressure": (1000, 1030),
        "uv": (1, 5)
    },

    "spring": {
        "humidity": (25, 65),
        "wind": (5, 30),
        "pressure": (995, 1025),
        "uv": (3, 8)
    },

    "summer": {
        "humidity": (30, 85),
        "wind": (5, 35),
        "pressure": (990, 1020),
        "uv": (6, 11)
    },

    "autumn": {
        "humidity": (30, 70),
        "wind": (0, 25),
        "pressure": (995, 1025),
        "uv": (2, 7)
    }
}


# 3. Identify season using month

def get_season(month):

    if month in [12, 1, 2]:
        return "winter"

    elif month in [3, 4]:
        return "spring"

    elif month in [5, 6, 7, 8]:
        return "summer"

    else:
        return "autumn"


# 4. Generate realistic synthetic weather

def generate_weather(city, current_date):

    month = current_date.month

    season = get_season(month)

    profile = city_climate[city]

    seasonal = season_data[season]

    # Generate temperature
    min_temp, max_temp = profile["temperature"][season]

    temperature = random.randint(min_temp, max_temp)

    # Generate rainfall
    rain_chance = profile["rain_probability"][season]

    if random.random() < rain_chance:

        rain = round(random.uniform(0.5, 60), 1)

    else:

        rain = 0.0

    # Generate humidity
    min_humidity, max_humidity = seasonal["humidity"]

    humidity = random.randint(min_humidity, max_humidity)

    humidity += profile["humidity_offset"]

    # Rainfall increases humidity
    if rain > 0:
        humidity += 10

    humidity = max(10, min(100, humidity))

    # Generate wind
    min_wind, max_wind = seasonal["wind"]

    wind = random.randint(min_wind, max_wind)

    wind += profile["wind_offset"]

    wind = max(0, min(60, wind))

    # Generate atmospheric pressure
    min_pressure, max_pressure = seasonal["pressure"]

    pressure = round(
        random.uniform(min_pressure, max_pressure), 1
    )

    # Generate visibility
    visibility = random.uniform(10, 20)

    # Rain reduces visibility
    if rain > 0:
        visibility -= min(rain * 0.2, 8)

    visibility = max(0.5, min(20, visibility))

    visibility = round(visibility, 1)

    # Generate UV index
    min_uv, max_uv = seasonal["uv"]

    uv_index = random.randint(min_uv, max_uv)

    # Rainy weather generally reduces UV exposure
    if rain > 0:
        uv_index = max(0, uv_index - 2)

    # Store all weather parameters
    weather = {

        "date": current_date.strftime("%Y-%m-%d"),

        "temperature_C": temperature,

        "humidity_pct": humidity,

        "wind_kmh": wind,

        "rain_mm": rain,

        "pressure_hpa": pressure,

        "visibility_km": visibility,

        "uv_index": uv_index
    }

    return weather


# 5. Generate 365 records for all cities

def generate_dataset():

    dataset = {}

    start_date = date(YEAR, 1, 1)

    for city in city_climate:

        city_data = []

        for day in range(365):

            current_date = start_date + timedelta(days=day)

            weather = generate_weather(city, current_date)

            city_data.append(weather)

        dataset[city] = city_data

    return dataset


# 6. Display all weather records of a selected city
def display_city_weather(city):
    if city not in dataset:
        print("City not found!")
        return

    print("\nWeather Data for:", city)
    print("-" * 120)

    print(
        f"{'Date':<12}"
        f"{'Temp(C)':<12}"
        f"{'Humidity(%)':<15}"
        f"{'Wind(km/h)':<15}"
        f"{'Rain(mm)':<12}"
        f"{'Pressure(hPa)':<17}"
        f"{'Visibility(km)':<17}"
        f"{'UV Index':<10}"
    )

    print("-" * 120)

    for record in dataset[city]:
        print(
            f"{record['date']:<12}"
            f"{record['temperature_C']:<12}"
            f"{record['humidity_pct']:<15}"
            f"{record['wind_kmh']:<15}"
            f"{record['rain_mm']:<12}"
            f"{record['pressure_hpa']:<17}"
            f"{record['visibility_km']:<17}"
            f"{record['uv_index']:<10}"
        )

# 7. Generate the complete dataset

dataset = generate_dataset()


# 8. Display available cities

print("\nPAKISTAN SYNTHETIC WEATHER DATASET")

print("\nAvailable Cities:")

for city in dataset:
    print("-", city)


# 9. Take input from user

selected_city = input("\nEnter city name: ").strip().title()

# Case-insensitive city matching
matched_city = None

for city in dataset:

    if city.lower() == selected_city.lower():

        matched_city = city

        break


# 10. Display selected city data

if matched_city is not None:

    display_city_weather(selected_city)

else:

    print("Invalid city name!")


# 11. Display dataset statistics
print("\nDataset Verification")

for city in dataset:
    print(city, ":", len(dataset[city]), "records")

print("\nTotal Cities:", len(dataset))
print("Total Weather Records:", sum(len(dataset[city]) for city in dataset))
```
